                                                   Module 1 – Data Loader

A DataLoader is a Python class used to read data from different types of files.
Instead of writing separate file-reading code every time, we create a DataLoader class with different methods for reading files.


Why do we use DataLoader?

Suppose we have three different files:
1.student.csv
2.student.json
3.student.txt

Without a DataLoader, we would write separate code every time we want to read them.
With a DataLoader, we create the class once and use its methods whenever required.

DataLoader
     |
     |--- read_csv()
     |--- read_json()
     |--- read_txt()
     |--- validate_path()
     |--- check_format()

2.VALIDATE FILE PATH
Before reading a file, we should check whether the file actually exists.

```python
import os

class DataLoader:

    def validate_path(self, file_path):
        if not os.path.isfile(file_path):
            raise FileNotFoundError("File not found")
```

First we need import os 
↓
Python module used for operating-system related tasks

os.path.isfile()
↓
Checks whether the given path is a file

raise FileNotFoundError()
↓
Generates an error if the file does not exist

```python
loader.validate_path("student.csv")

#If the file exists, nothing happens.

#If the file does not exist:

FileNotFoundError: File not found
```

3.READING CSV FILES

CSV files are commonly used for storing tabular data such as:

Name,Age,Marks
Ajay,23,99
Kumar,22,95

Python uses Pandas to easily read CSV files.

```python
import pandas as pd

class DataLoader:

    def read_csv(self, file_path):
        return pd.read_csv(file_path)
```

Example:

loader.read_csv("student.csv")

    Name  Age  Marks
0   Ajay   23     99
1  Kumar   22     95

The returned data is a DataFrame.

4.READING JSON FILES

JSON commonly stores data in key-value format.

Example:
{
    "Name": "Kiran",
    "Age": 20
}

Python provides the json module to read JSON files.

```python
import json

class DataLoader:

    def read_json(self, file_path):
        with open(file_path, "r") as file:
            return json.load(file)
```

Explanation:

open()
↓
Opens the file

"r"
↓
Read mode

json.load()
↓
Reads JSON data
↓
Converts it into a Python object

Example:

loader.read_json("student.json")

Output:

{'Name': 'Kiran', 'Age': 20}

5.READING TEXT FILES

A TXT file contains normal text.

Example:

Welcome to Python
This is a text file

We can read it using open().

```python
class DataLoader:

    def read_txt(self, file_path):
        with open(file_path, "r") as file:
            return file.read()
```

open(file_path, "r")
↓
Opens the text file in read mode

file.read()
↓
Reads the complete content

return
↓
Returns the file content

```python
Example:

loader.read_txt("student.txt")

Output:

Welcome to Python
This is a text file
```

6.CHECKING FILE FORMAT

Sometimes the user may give an unsupported file such as:

student.pdf
student.xlsx
student.exe

Our DataLoader only supports:

.csv
.json
.txt

Our DataLoader only supports:

.csv
.json
.txt

```python
import os

class DataLoader:

    def check_format(self, file_path):
        extension = os.path.splitext(file_path)[1].lower()

        if extension not in [".csv", ".json", ".txt"]:
            raise ValueError("Unsupported file format")
```

```python
Explanation:

os.path.splitext() separates the filename and extension.

Example:

os.path.splitext("student.csv")

Gives:

("student", ".csv")

Then:

[1]

gets the extension:

.csv

lower() converts the extension into lowercase.

So:

.CSV
.csv
.Csv

will all become:

.csv

Then we check whether it is supported.

If the extension is not:

.csv
.json
.txt

we raise:

ValueError: Unsupported file format
```
