```python
import os
import json
import pandas as pd
```

```python

class DataLoader:

    # Checks whether the given file exists
    def validate_path(self, file_path):
        if not os.path.isfile(file_path):
            raise FileNotFoundError("File not found")

    # Reads a CSV file and returns a DataFrame
    def read_csv(self, file_path):
        return pd.read_csv(file_path)

    # Reads a JSON file and returns the data
    def read_json(self, file_path):
        with open(file_path, "r") as file:
            return json.load(file)

    # Reads a TXT file and returns its content
    def read_txt(self, file_path):
        with open(file_path, "r") as file:
            return file.read()

    # Checks whether the file format is supported
    def check_format(self, file_path):
        extension = os.path.splitext(file_path)[1].lower()

        if extension not in [".csv", ".json", ".txt"]:
            raise ValueError("Unsupported file format")
```

```python
# Creates an object of the DataLoader class
loader = DataLoader()

# Tests the file reading methods
loader.read_csv("sample.csv")
loader.read_json("sample.json")
loader.read_txt("sample.txt")
loader.check_format("sample.txt")
```
