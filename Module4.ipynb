1) WHAT IS EXCEPTION HANDLING?

Exception handling is used to handle errors that occur while a Python program is running.

For example, if we try to open a file that doesn't exist, Python will generate an error.    


Instead of allowing the program to stop suddenly, we can handle the error using
try
except
finally
raise

1) CUSTOM EXCEPTION

A custom exception is an exception created by the programmer for a specific purpose.

We create a custom exception by inheriting from Python's built-in Exception class.

```python
class InvalidFileException(Exception):
    """Custom exception for invalid files."""
    pass

#InvalidFileException can now be used to represent an invalid file error.
```

class → Used to create a class.
InvalidFileException → Name of our custom exception.
Exception → Built-in Python exception class that we inherit from.
pass → Means there is no additional code inside the class.

2) TRY

The try block contains the code that may cause an exception.

```python
try:
    file_path = "student.pdf"

    if not file_path.endswith((".csv", ".json", ".txt")):
        raise InvalidFileException("Unsupported file format")
```

3) RAISE

The raise keyword is used to manually generate an exception.

```python
raise InvalidFileException("Unsupported file format")
```

4) EXCEPT

The except block is used to handle an exception that occurs inside the try block.

```python
except InvalidFileException as error:
    print("Error:", error)
```

5) FINALLY

The finally block always executes, whether an exception occurs or not.

```python
finally:
    print("File validation completed")
```
