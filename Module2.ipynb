1.WRITING CSV FILES

CSV files are used to store tabular data in rows and columns.

For example:

Name,Age,Marks
Ajay,23,99
Kumar,22,95

We use Pandas to write a DataFrame into a CSV file.

```python
def write_csv(self, data, file_path):
    data.to_csv(file_path)
```

2) WRITING JSON FILES

JSON stores data mainly in key-value format.

Example:

{
    "name": "Kiran",
    "age": 20
}

```python
def write_json(self, data, file_path):
    with open(file_path, "w") as file:
        json.dump(data, file, indent=4)


#Here, indent=4 means use 4 spaces for each level of indentation.
```

Explanation:

open()
↓
Opens the file

"w"
↓
Write mode

json.dump()
↓
Converts Python data into JSON
and writes it to the file

indent=4
↓
Makes JSON properly formatted and readable

3) WRITING PICKLE FILES

Pickle is used to save Python objects directly into a file.

```python
def write_pickle(self, data, file_path):
    with open(file_path, "wb") as file:
        pickle.dump(data, file)
```

Explanation:
"wb"
↓
Write Binary mode

pickle.dump()
↓
Python Object → Pickle File

```python
4) WRITING TEXT FILES

TXT files store normal text.
```

```python
def write_txt(self, data, file_path):
    with open(file_path, "w") as file:
        file.write(data)
```

Explanation:

"w"
↓
Write mode

file.write()
↓
Writes the given text into the file
