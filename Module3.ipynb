1) WHAT IS LOGGING?

Logging means recording information about what is happening inside an application.

For example, our application may:

Start successfully
Receive a warning
Encounter an error

Instead of only using print(), we can use logging to record these events in a log file.

Our requirement is to create:

logs/application.log

2) IMPORT LOGGING

Python provides a built-in logging module.

```python
import loggin
```

3) CREATING LOG DIRECTORY

Before writing the log file, we need the logs folder.

```python
import os

os.makedirs("logs", exist_ok=True)
```

Why exist_ok=True?

If the logs folder already exists, Python won't give an error.


logs doesn't exist
        ↓
Create logs folder

logs already exists
        ↓
Continue without error

```python
4) CONFIGURING LOGGING

We use logging.basicConfig() to configure the logger.
```

```python
logging.basicConfig(
    filename="logs/application.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)
```

```python
#filename
filename="logs/application.log"

#Specifies where the logs should be stored.
```

So our logs will be saved inside:

logs/
    application.log

```python
#level

level=logging.INFO

This tells Python to record:

INFO
WARNING
ERROR
```

```python
#format

format="%(asctime)s - %(levelname)s - %(message)s"
```

Controls how each log entry will look.

For example:

2026-09-10 12:00:10,123 - INFO - Application started
2026-09-10 12:00:15,456 - WARNING - File is almost empty
2026-09-10 12:00:20,789 - ERROR - File not found

5) LOGGING LEVELS

Python provides different logging levels.

For this task, we need:

INFO
WARNING
ERROR


INFO

Used for normal information about the application.

```python
#Example:

logger.info("Application started")

#Meaning:
#The application has started normally.
```

WARNING

Used when something unexpected happens, but the application can continue.

```python
#Example:

logger.warning("File is almost empty")

#Meaning:
#Something may need attention, but it is not a serious error.
```

ERROR

Used when something goes wrong.

```python
logger.error("File not found")
```

Meaning:

An error occurred while running the application.
